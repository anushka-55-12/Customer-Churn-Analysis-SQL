# Customer Churn Analysis - Feature Engineering

## Objective

The purpose of feature engineering is to transform raw customer information into meaningful variables that can improve the performance of predictive models.

## Business Context

Customer churn is influenced by several factors such as customer tenure, contract type, monthly spending, and service subscriptions.

Creating engineered features allows the model to better capture customer behavior patterns and identify customers at risk of churning.

## Goals

- Create meaningful business-driven features
- Convert categorical variables into machine-readable format
- Prepare the dataset for predictive modeling
- Improve model performance and interpretability

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/cleaned/churn_cleaned.csv")

df.head()
```

## Feature 1: Average Monthly Customer Value

### Business Objective

Estimate the average amount spent by a customer during their relationship with the company.

### Business Rationale

Customers generating higher revenue may exhibit different churn behavior compared to low-value customers.

```python
df["AvgSpendPerMonth"] = (
    df["TotalCharges"] /
    (df["tenure"] + 1)
)
```

## Feature 2: New Customer Indicator

### Business Objective

Identify customers who joined recently.

### Business Rationale

New customers typically have higher churn rates because they have not yet developed loyalty toward the service.

```python
df["NewCustomer"] = np.where(
    df["tenure"] <= 12,
    1,
    0
)
```

## Feature 3: High Monthly Charge Indicator

### Business Objective

Identify customers paying above-average monthly fees.

### Business Rationale

Pricing sensitivity is a common driver of customer churn.

```python
avg_charge = df["MonthlyCharges"].mean()

df["HighMonthlyCharge"] = np.where(
    df["MonthlyCharges"] > avg_charge,
    1,
    0
)
```

## Feature 4: Service Adoption Score

### Business Objective

Measure the number of services subscribed to by each customer.

### Business Rationale

Customers using multiple services tend to be more engaged and may be less likely to churn.

```python
service_cols = [
    'PhoneService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

for col in service_cols:
    df[col] = np.where(df[col] == "Yes", 1, 0)

df["TotalServices"] = df[service_cols].sum(axis=1)
```

```python
df.to_csv(
    "../data/cleaned/churn_featured.csv",
    index=False
)
```
